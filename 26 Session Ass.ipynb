```python
#1.Prepare a sequence dataset from Flipkart product prices over 10 days (you can create fake data) and reshape it into 
# the format required for LSTM input using NumPy.

import numpy as np

prices = np.array([999, 1049, 1099, 1079, 1149, 1199, 1249, 1229, 1299, 1349], dtype=float)

X = []
y = []

for i in range(len(prices) - 3):
    X.append(prices[i:i+3])
    y.append(prices[i+3])


X = np.array(X)
y = np.array(y)

X = X.reshape((X.shape[0], X.shape[1], 1))

print("Price sequence:", prices)
print("X shape:", X.shape)
print("y shape:", y.shape)
print("X:")
print(X)
print("y:")
print(y)
```

```python
# 2.Build a simple LSTM model in Keras to predict the next value in your prepared Flipkart price sequence and train it for 10 epochs.

import numpy as np 
import tensorflow as tf 
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense

prices = np.array([999, 1049, 1099, 1079, 1149, 1199, 1249, 1229, 1299, 1349], dtype=float)

X = []
y = []

for i in range(len(prices) - 3):
    X.append(prices[i:i+3])
    y.append(prices[i+3])


X = np.array(X)
y = np.array(y)

X = X.reshape((X.shape[0], X.shape[1], 1))

model = Sequential([LSTM(32, input_shape=(3,1)), Dense(1)]) 

model.compile(optimizer="adam", loss="mse")
history = model.fit(X, y, epochs=10, verbose=1) 
```

```python
# 3.Evaluate your trained LSTM model by predicting the next price in the sequence, and print both the predicted and actual values for comparison.
# <br><br><em><strong>Hint:</strong> Use model.predict() and compare with the last actual value in your test set.</em>

last_sequence = np.array([1199, 1249, 1229], dtype=float)
last_sequence = last_sequence.reshape((1, 3, 1))

predicted_price = model.predict(last_sequence)

actual_price = 1299

print("Pridicted Price:", predicted_price[0][0])
print("Actual Price:", actual_price)
```
