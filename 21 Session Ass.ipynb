```python
# 1.Create a new Keras Sequential model and add a Conv2D layer with 32 filters, a 3x3 kernel, 
# and ReLU activation for image input shape (32, 32, 3) as used in CIFAR-10.

import numpy as np 
import tensorflow as tf
from tensorflow.keras import Sequential 
from tensorflow.keras.layers import Conv2D, MaxPooling2D, BatchNormalization, Flatten, Dropout, Dense


model = Sequential()

model.add(Conv2D(filters=32, kernel_size=(3,3), activation="relu", input_shape=(32, 32, 3)))

model.summary()
```

```python
#2.Add a MaxPooling2D layer after your Conv2D layer to reduce spatial dimensions, 
# then flatten the output and add a Dense layer with 64 units and ReLU activation.

import tensorflow as tf 
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense

model = Sequential([ 
    Conv2D(32, (3, 3), activation="relu", input_shape=(64, 64, 3)),

    MaxPooling2D(pool_size=(2,2)),

    Flatten(),

    Dense(64, activation="relu")
    ])

model.summary()
```

```python
# 3.Insert a BatchNormalization layer after the Conv2D and before MaxPooling2D, and add a Dropout layer (rate 0.3) after flattening to help prevent overfitting.<br><br><em><strong>Hint:
# </strong> BatchNorm stabilizes training, Dropout randomly disables neurons during training.</em>\

model = Sequential([
    Conv2D(32, (3,3), activation="relu", input_shape=(64, 64, 3)),

    BatchNormalization(),
    MaxPooling2D(pool_size=(2,2)),
    Flatten(),
    Dropout(0.3),

    Dense(64, activation="relu")
])

model.summary()
```

```python
# 4.
Compile your model using the 'adam' optimizer, 'categorical_crossentropy' loss, and track 'accuracy' as a metric. Then, train the model for 5 epochs using random data shaped like CIFAR-10 images and labels.<br><br><em><strong>Constraint:</strong> Use numpy to generate random input data and one-hot encoded labels for testing if you don't want to download the full CIFAR-10 dataset.</em>

model = Sequential([
    Conv2D(32, (3, 3), activation='relu', input_shape=(32, 32, 3)),
    BatchNormalization(),
    MaxPooling2D(pool_size=(2,2)),
    Flatten(),
    Dropout(0.3),
    Dense(64, activation="relu"),
    Dense(10, activation="softmax")
])


model.compile(optimizer="adam", loss="categorical_crossentropy", metrics=["accuracy"])

X_train = np.random.random((1000, 32, 32, 3))

y_labels = np.random.randint(0, 10, 1000) 

y_train = np.eye(10)[y_labels]

model.fit(X_train, y_train, epochs=5, batch_size=32)
```

```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, BatchNormalization
from tensorflow.keras.layers import MaxPooling2D, Flatten, Dense, Dropout

model = Sequential([
    Conv2D(32, (3, 3), activation='relu', input_shape=(32, 32, 3)),
    BatchNormalization(),
    MaxPooling2D(pool_size=(2, 2)),


    Conv2D(64, (3,3), activation="relu"),
    BatchNormalization(),
    MaxPooling2D(pool_size=(2,2)),

    Flatten(),
    Dropout(0.3),
    Dense(64, activation="relu"),
    Dense(10, activation="softmax")
])

model.compile(optimizer="adam", loss="categorical_crossentropy", metrics=["accuracy"])

X_train = np.random.random((1000, 32, 32, 3))

y_labels = np.random.randint(0, 10, 1000)

y_train = np.eye(10)[y_labels]


%history = model.fit(X_train, y_train, epochs=5, batch_size=32)

model.summary()
```
