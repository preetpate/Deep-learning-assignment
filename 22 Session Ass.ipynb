```python
# 1.Load any 3 sample images (your choice: selfies, food, cricket, or memes) and use OpenCV or PIL to resize each to 224x224 pixels,
# saving the resized versions in a new folder.

from PIL import Image
import os

os.makedirs("resized_images", exist_ok=True)

image_files = [
    "Image1.jpg",
    "Image2.jpg",
    "Image3.jpg"
]

for i, file in enumerate(image_files, start=1):
    img = Image.open(file)

    resized_img = img.resize((224,224))

    resized_img.save(f"resized_images/rresized_{i}.jpg")

    print(f"Saved: resized_images/resized_{i}.jpg")

print("All images resized successfully ")
```

```python
# 2.Convert a colored image (choose any photo you like) to grayscale and then back to RGB using Python libraries.
# Display all three versions (original, grayscale, RGB) using matplotlib.

from PIL import Image
import matplotlib.pyplot as plt 

img = Image.open("Image1.jpg").convert("RGB")

gray_img = img.convert("L")

rgb_img = gray_img.convert("RGB")

plt.figure(figsize=(12,4))

plt.subplot(1, 3, 1)
plt.imshow(img)
plt.title("Original Animals Images")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(gray_img, cmap="gray")
plt.title("Grayscale Images")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(rgb_img)
plt.title("RGB Image")
plt.axis("off")  

plt.show()
```

```python
# 3.Use Keras' ImageDataGenerator to apply random rotation and horizontal flip to an image of your choice (for example, a Zomato food pic or an IPL player). Generate and display 4 augmented versions.<br><br><em><strong>Hint:</strong>
# Set rotation_range and horizontal_flip in ImageDataGenerator, then use flow() or flow_from_directory().</em>

import numpy as np 
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.utils import img_to_array, load_img


img = load_img("Image2.jpg")
img = img.resize((224,224))

x = img_to_array(img)
x = np.expand_dims(x, axis=0)

datagen = ImageDataGenerator(rotation_range=30, horizontal_flip=True)
augmented_images = datagen.flow(x, batch_size=1)

plt.figure(figsize=(10,8))

for i in range (4):
    batch = next(augmented_images)
    aug_img = batch[0].astype("uint8")

    plt.subplot(2, 2, i+1)
    plt.imshow(aug_img)
    plt.title(f"Augmented Image {i+1}")
    plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
# 4.Write code to increase and decrease the brightness of a selfie or any image by 40% using ImageDataGenerator's brightness_range parameter. 
# Show the original, brightened, and darkened images side by side

from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.utils import img_to_array, load_img

img = load_img("Image3.jpg")
img = img.resize((224,224))

x = img_to_array(img)
x = np.expand_dims(x, axis=0)

bright_datagen = ImageDataGenerator(brightness_range=[1.4, 1.4])
bright_generator = bright_datagen.flow(x, batch_size=1)

bright_img = next(bright_generator)[0].astype("uint8")

dark_datagen = ImageDataGenerator(brightness_range=[0.6, 0.6])
dark_generator = dark_datagen.flow(x, batch_size=1)

dark_img = next(dark_generator)[0].astype("uint8")

plt.figure(figsize=(12,4))

plt.subplot(1, 3, 1)
plt.imshow(x[0].astype("uint8"))
plt.title("Original")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(bright_img)
plt.title("Brightened (+40%)")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(dark_img)
plt.title("Darkened (-40%)")
plt.axis("off")

plt.tight_layout()
plt.show
```

```python
#5.Pick any meme or cricket image and use ImageDataGenerator to apply a shear transformation. 
# Explain in one line how the image changes visually after shearing.

img = load_img("Image3.jpg")
img = img.resize((224,224))

x = img_to_array(img)
x = np.expand_dims(x, axis=0)

datagen = ImageDataGenerator(shear_range=30)
generator = datagen.flow(x, batch_size=1)

sheared_img = next(generator)[0].astype("uint8") 

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(x[0].astype("uint8"))
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(sheared_img)
plt.title("Sheared Image")
plt.axis("off")

plt.tight_layout()
plt.show()
```
